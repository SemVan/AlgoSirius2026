```python
class Node:
    def __init__(self, data, next_ptr=None):
        self.data = data
        self.next = next_ptr
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        if self.head is None:
            self.head = Node(data)
            return
        curnode = self.head
        while curnode.next is not None:
            curnode = curnode.next
        curnode.next = Node(data)

    def print(self):
        curnode = self.head
        while curnode is not None:
            print(curnode.data)
            curnode = curnode.next

    def insert(self, num, data):
        if num == 0:
            self.head = Node(data, self.head)
            return
        curnode = self.head
        step = 0
        while step < num-1 and curnode is not None:
            step += 1
            curnode = curnode.next
        if step < num-1 or curnode is None:
            print('num must be less than length of list')
            return
        curnode.next = Node(data, curnode.next)
        return
        

    def delete(self, num):
        if self.head:
            if num == 0:
                self.head = self.head.next
                return
            curnode = self.head
            step = 0
            while step < num-1 and curnode is not None:
                step += 1
                curnode = curnode.next
            if curnode is None or curnode.next is None:
                print('num must be less than length of list')
                return
            curnode.next = curnode.next.next

    def find(self, data):
        curnode = self.head
        step = 0
        while curnode is not None:
            if curnode.data == data:
                return step
            step += 1
            curnode = curnode.next
        return -1
        
            
```
