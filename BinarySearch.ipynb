```python
a = [1, 5, 9, 17, 34, 36, 42, 54, 58, 62, 78]

def binary_search(array, value):
    left = 0
    right = len(array) - 1
    cnt = 0
    while left <= right:
        cnt += 1
        mid = (right + left) // 2
        if value < array[mid]:
            right = mid - 1
        elif  value > array[mid]:
            left = mid + 1
        else:
            return mid, cnt
    return -1, cnt
#[1,2,3]
def binary_search_rec(array, value, cnt=0):
    cnt += 1
    if len(array) == 0:
        return -1, cnt
    mid = len(array) // 2
    if value < array[mid]:
        return binary_search_rec(array[:mid], value, cnt)
    elif  value > array[mid]:
        res, cnt = binary_search_rec(array[mid+1:], value, cnt)
        if res != -1:
            return res + 1 + mid, cnt
        else:
            return -1, cnt     
    else:
        return mid, cnt


def linear_search(array, value):
    for i in range(len(array)):
        if array[i] == value:
            return i, i+1
    return -1, len(array)
```

```python
import numpy as np
import time


lens = [10, 100, 500, 1000, 5000, 10000, 50000, 100000, 500000, 1000000]
cnts_binary = []
cnts_linear = []
times_binary = []
times_linear = []
for l in lens:
    cnts_binary_tmp = []
    cnts_linear_tmp = []
    times_binary_tmp = []
    times_linear_tmp = []
    for i in range(10):
        a = sorted(np.random.randint(0, l*2, l))
        val = np.random.randint(0, l*2, 1)
        start = time.time()
        _, cnt1 = binary_search_rec(a, val)
        times_binary_tmp.append(time.time() - start)
        start = time.time()
        _, cnt2 = linear_search(a, val)
        times_linear_tmp.append(time.time() - start)
        cnts_binary_tmp.append(cnt1)
        cnts_linear_tmp.append(cnt2)
    cnts_binary.append(np.mean(cnts_binary_tmp))
    cnts_linear.append(np.mean(cnts_linear_tmp))
    times_binary.append(np.mean(times_binary_tmp))
    times_linear.append(np.mean(times_linear_tmp))
```
